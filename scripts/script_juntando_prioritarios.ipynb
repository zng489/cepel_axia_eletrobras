```python
import pandas as pd

files = [
    r"C:\Users\Yuan\Desktop\git_hub\cepel_axia_eletrobras\scripts\doc_2_.csv",
    r"C:\Users\Yuan\Desktop\git_hub\cepel_axia_eletrobras\scripts\doc_3_.csv",
    r"C:\Users\Yuan\Desktop\git_hub\cepel_axia_eletrobras\scripts\doc_4_.csv"
]

dfs = [
    pd.read_csv(f, sep=',', encoding='utf-8')
    for f in files
]

df = pd.concat(dfs, ignore_index=True)

# ---------------------------
# FILTER 1 (exclusão)
# ---------------------------
excluir = [
    "ANALISE_VISUALIZACAO_DADOS_PRIORITARIO",
    "ARMAZENAMENTO_E_GESTAO_DADOS_PRIORITARIO",
    "ARQUITETURA_DADOS_PRIORITARIO",
    "BANCO_DADOS_NOSQL_PRIORITARIO",
    "BANCO_DADOS_OLAP_PRIORITARIO",
    "BANCO_DADOS_SQL_PRIORITARIO",
    "BANCO_DADOS_VETORIAL_PRIORITARIO",
    "CONFORMIDADE_DADOS_PRIORITARIO",
    "CONTROLE_VERSAO_PRIORITARIO",
    "FORMATO_ARMAZENAMENTO_PRIORITARIO",
    "GOVERNANCA_GESTAO_DADOS_PRIORITARIO",
    "INGESTAO_DADOS_PRIORITARIO",
    "LINGUAGEM_PRIORITARIA",
    "MONITORAMENTO_DADOS_PRIORITARIO",
    "ORQUESTRACAO_DADOS_PRIORITARIO",
    "SEGURANCA_DADOS_PRIORITARIO",
    "SISTEMA_OPERACIONAL_PRIORITARIO",
    "TIPO_DADOS_PRIORITARIO",
    "TRANSFORMACAO_DADOS_PRIORITARIO",
    "VIRTUALIZACAO_PRIORITARIO"
]

df = df[~df["CATEGORIA"].isin(excluir)]

# ---------------------------
# FILTER 2
# ---------------------------
categorias_validas = [
    "BOAS_PRATICAS_ENGENHARIA",
    "CI_CD",
    "DOCUMENTACAO_COLABORACAO",
    "FERRAMENTAS_GESTAO_PROJETOS",
    "METODOLOGIAS_GESTAO"
]

df = df[
    (df["RESPOSTA"] != "") &
    (df["CATEGORIA"].isin(categorias_validas))
]

# ---------------------------
# RENAME COLUMNS
# ---------------------------
df = df.rename(columns={
    "SENIORIDADE_TEMPO_DE_CASA_NO_CEPEL": "Serioridade no Cepel",
    "SENIORIDADE_TEMPO_VOCE_JA_PROGRAMA_DESENVOLVE_SISTEMAS": "Senioridade em Dev.Sistemas",
    "SENIORIDADE_TEMPO_VOCE_JA_TRABALHA_COM_DADOS_GOVERNANCA_ENGENHARIA_CIENCIA_OU_ANALISE_DE_DADOS": "Senioridade em Dados"
})

# ---------------------------
# REPLACE VALUES
# ---------------------------

# Cepel
df["Serioridade no Cepel"] = df["Serioridade no Cepel"].replace({
    "Senior": "Senior → 6 anos ou mais",
    "Pleno": "Pleno → 3-5 anos",
    "Junior": "Junior → Menos de 1 ano ou 1-2 anos"
})

# Dev
df["Senioridade em Dev.Sistemas"] = df["Senioridade em Dev.Sistemas"].replace({
    "Senior": "Senior → 5 anos ou mais",
    "Pleno": "Pleno → 3-5 anos",
    "Junior": "Junior → Menos de 1 ano ou 1-2 anos"
})

# Dados
df["Senioridade em Dados"] = df["Senioridade em Dados"].replace({
    "Senior": "Senior → 5 anos ou mais",
    "Pleno": "Pleno → 3-5 anos",
    "Junior": "Junior → Menos de 1 ano ou 1-2 anos"
})
```

```python
import pandas as pd

files = [
    r"C:\Users\Yuan\Desktop\git_hub\cepel_axia_eletrobras\scripts\doc_2_.csv",
    r"C:\Users\Yuan\Desktop\git_hub\cepel_axia_eletrobras\scripts\doc_3_.csv",
    r"C:\Users\Yuan\Desktop\git_hub\cepel_axia_eletrobras\scripts\doc_4_.csv"
]

dfs = [
    pd.read_csv(f, sep=',', encoding='utf-8')
    for f in files
]

df = pd.concat(dfs, ignore_index=True)
# ---------------------------
# FILTER CATEGORIA (IN)
# ---------------------------
categorias = [
    "ANALISE_VISUALIZACAO_DADOS_PRIORITARIO",
    "ARMAZENAMENTO_E_GESTAO_DADOS_PRIORITARIO",
    "ARQUITETURA_DADOS_PRIORITARIO",
    "BANCO_DADOS_NOSQL_PRIORITARIO",
    "BANCO_DADOS_OLAP_PRIORITARIO",
    "BANCO_DADOS_SQL_PRIORITARIO",
    "BANCO_DADOS_VETORIAL_PRIORITARIO",
    "CONFORMIDADE_DADOS_PRIORITARIO",
    "CONTROLE_VERSAO_PRIORITARIO",
    "FONTE_DADOS_PRIORITARIO",
    "FORMATO_ARMAZENAMENTO_PRIORITARIO",
    "GOVERNANCA_GESTAO_DADOS_PRIORITARIO",
    "IDES_PRIORITARIO",
    "INGESTAO_DADOS_PRIORITARIO",
    "LINGUAGEM_PRIORITARIA",
    "MONITORAMENTO_DADOS_PRIORITARIO",
    "ORQUESTRACAO_DADOS_PRIORITARIO",
    "SEGURANCA_DADOS_PRIORITARIO",
    "SISTEMA_OPERACIONAL_PRIORITARIO",
    "TIPO_DADOS_PRIORITARIO",
    "TRANSFORMACAO_DADOS_PRIORITARIO",
    "VIRTUALIZACAO_PRIORITARIO"
]

df = df[df["CATEGORIA"].isin(categorias)]

# ---------------------------
# SPLIT + EXPLODE (Power Query Split + ExpandListColumn)
# ---------------------------
df["RESPOSTA"] = df["RESPOSTA"].fillna("")
df["RESPOSTA"] = df["RESPOSTA"].str.split(";")

df = df.explode("RESPOSTA")

# remover vazios
df = df[df["RESPOSTA"].str.strip() != ""]

# ---------------------------
# RENAME
# ---------------------------
df = df.rename(columns={
    "SENIORIDADE_TEMPO_DE_CASA_NO_CEPEL": "Serioridade no Cepel",
    "SENIORIDADE_TEMPO_VOCE_JA_PROGRAMA_DESENVOLVE_SISTEMAS": "Senioridade em Dev.Sistemas",
    "SENIORIDADE_TEMPO_VOCE_JA_TRABALHA_COM_DADOS_GOVERNANCA_ENGENHARIA_CIENCIA_OU_ANALISE_DE_DADOS": "Senioridade em Dados"
})

# ---------------------------
# REPLACE
# ---------------------------
df["Serioridade no Cepel"] = df["Serioridade no Cepel"].replace({
    "Senior": "Senior → 6 anos ou mais",
    "Pleno": "Pleno → 3-5 anos",
    "Junior": "Junior → Menos de 1 ano ou 1-2 anos"
})

df["Senioridade em Dev.Sistemas"] = df["Senioridade em Dev.Sistemas"].replace({
    "Senior": "Senior → 5 anos ou mais",
    "Pleno": "Pleno → 3-5 anos",
    "Junior": "Junior → Menos de 1 ano ou 1-2 anos"
})

df["Senioridade em Dados"] = df["Senioridade em Dados"].replace({
    "Senior": "Senior → 5 anos ou mais",
    "Pleno": "Pleno → 3-5 anos",
    "Junior": "Junior → Menos de 1 ano ou 1-2 anos"
})

# ---------------------------
# DUPLICATE + SWAP (parte final do Power Query)
# ---------------------------

# duplicar colunas
df["TECNOLOGIA - Copy"] = df["TECNOLOGIA"]
df["RESPOSTA - Copy"] = df["RESPOSTA"]

# remover TECNOLOGIA original
df = df.drop(columns=["TECNOLOGIA"])

# renomear RESPOSTA -> TECNOLOGIA
df = df.rename(columns={"RESPOSTA": "TECNOLOGIA"})

df_p_1 = df
```
